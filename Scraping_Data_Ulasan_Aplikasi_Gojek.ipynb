```python
from google_play_scraper import app, reviews, Sort
scrapreview, contunuation_token = reviews(
    'com.gojek.app',
    lang='id',
    country='id',
    sort=Sort.MOST_RELEVANT,
    count=12000
)
```

```python
import pandas as pd
df_reviews = pd.DataFrame(scrapreview)
```

```python
def tentukan_kategori(text):
    # Ubah teks menjadi huruf kecil untuk memudahkan pencocokan
    text_lower = str(text).lower()

    # Kumpulan kata kunci berdasarkan dimensi kuesioner
    kat_teknis = ['aplikasi', 'error', 'eror', 'update', 'lag', 'force close', 'blank', 'sistem', 'maps', 'titik', 'apk', 'lemot', 'bug']
    kat_waktu = ['lama', 'waktu', 'cepat', 'lambat', 'telat', 'menunggu', 'estimasi', 'jam', 'macet']
    kat_driver = ['driver', 'pengemudi', 'ojol', 'motor', 'mobil', 'ramah', 'sopan', 'ngebut', 'cancel', 'batal', 'njemput']
    kat_cs = ['cs', 'customer service', 'bantuan', 'keluhan', 'respon', 'tanggapan', 'refund', 'ai']
    kat_harga = ['mahal', 'murah', 'harga', 'promo', 'diskon', 'ongkir', 'tarif', 'voucher', 'bayar', 'gopay', 'saldo']

    # Logika pencocokan
    if any(word in text_lower for word in kat_driver):
        return 'Interaksi & Pelayanan Driver'
    elif any(word in text_lower for word in kat_teknis):
        return 'Kualitas Teknis & Aplikasi'
    elif any(word in text_lower for word in kat_waktu):
        return 'Ketepatan Waktu & Efisiensi'
    elif any(word in text_lower for word in kat_harga):
        return 'Harga & Promosi'
    elif any(word in text_lower for word in kat_cs):
        return 'Respons Pengembang & CS'
    else:
        return 'Lainnya / Umum'
```

```python
df_reviews['Kategori'] = df_reviews['content'].apply(tentukan_kategori)

print("\nContoh Hasil Kategorisasi:")
display(df_reviews[['content', 'Kategori', 'score']].head(10))
```

```python
df_export = df_reviews[['content', 'Kategori']].rename(columns={'content': 'Review'})
```

```python
file_name = 'ulasan_gojek_kategori.csv'
df_export.to_csv(file_name, index=False, encoding='utf-8')
```
